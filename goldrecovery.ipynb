# 1.2 Check the Recovery Formula Accuracy

```python
import numpy as np
from sklearn.metrics import mean_absolute_error

# Extract columns
C = train['rougher.output.concentrate_au']
F = train['rougher.input.feed_au']
T = train['rougher.output.tail_au']

# Calculate recovery
recovery_calc = (C * (F - T)) / (F * (C - T)) * 100

# Drop NaNs
mask = ~recovery_calc.isna() & ~train['rougher.output.recovery'].isna()
mae = mean_absolute_error(train.loc[mask, 'rougher.output.recovery'], recovery_calc[mask])
print("MAE between actual and calculated recovery:", mae)
```

The MAE is very small, confirming that the recovery column was correctly calculated.

# Step 3: Model Development

# 3.1 Define sMAPE Metric

```python
def smape(y_true, y_pred):
    return 100 * np.mean(2 * np.abs(y_pred - y_true) / (np.abs(y_true) + np.abs(y_pred)))

def final_smape(y_true, y_pred):
    s1 = smape(y_true[:, 0], y_pred[:, 0])
    s2 = smape(y_true[:, 1], y_pred[:, 1])
    return 0.25 * s1 + 0.75 * s2
```

# 3.2 Train and Evaluate Models

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_score
from sklearn.metrics import make_scorer
import numpy as np

# sMAPE function
def smape(y_true, y_pred):
    denominator = (np.abs(y_true) + np.abs(y_pred)) / 2
    diff = np.abs(y_true - y_pred) / denominator
    diff[denominator == 0] = 0.0
    return np.mean(diff) * 100

# Define scorer
smape_scorer = make_scorer(smape, greater_is_better=False)

# Prepare data
X = X_train
y_rougher = y_train['rougher.output.recovery']
y_final = y_train['final.output.recovery']

# Rougher model and scoring
model_rougher = RandomForestRegressor(n_estimators=100, random_state=42)
score_rougher = cross_val_score(model_rougher, X, y_rougher, cv=5, scoring=smape_scorer)
print(f'Mean cross-validated sMAPE (Rougher - RandomForest): {-score_rougher.mean():.2f}')

# Final model and scoring
model_final = RandomForestRegressor(n_estimators=100, random_state=42)
score_final = cross_val_score(model_final, X, y_final, cv=5, scoring=smape_scorer)
print(f'Mean cross-validated sMAPE (Final - RandomForest): {-score_final.mean():.2f}')

# Weighted average
weighted_smape = 0.25 * (-score_rougher.mean()) + 0.75 * (-score_final.mean())
print(f'Weighted average sMAPE (RandomForest): {weighted_smape:.2f}')



       
```

The Random Forest model demonstrated solid performance on both recovery targets:

Rougher output recovery: sMAPE ≈ 7.85

Final output recovery: sMAPE ≈ 8.78

Weighted sMAPE (25% rougher, 75% final): 8.55

The sMAPE values are relatively low, indicating that the model is making accurate predictions with minimal error on both sub-targets. Given the reliable cross-validation performance, Random Forest is a strong candidate for the final model selection.

# 3.3 Tune Random Forest
